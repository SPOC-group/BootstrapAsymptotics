```julia
using BootstrapAsymptotics
```

```julia
problem = BootstrapAsymptotics.EnsembledRidge(
    α = 10.0,
    Δ = 1.0,
    Δ̂ = 1.0,
    λ = 1.0,
    ρ = 1.0
)
```

```julia
BootstrapAsymptotics.state_evolution(
    problem,
    NoResampling(),
    NoResampling();
    rtol=1e-2,
    max_iteration=1,
    show_progress=true,
)
```
